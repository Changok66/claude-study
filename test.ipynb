```python
import pandas as pd
df = pd.read_csv("data/sales.csv")

# 전체 매출 총합
df["매출"] = df["수량"] * df["단가"]
전체합 = df["매출"].sum()
print("전체 매출:", 전체합)

# 지역별로 나눠서 더한 것의 합
지역별합 = df.groupby("지역")["매출"].sum().sum()
print("지역별 합의 합:", 지역별합)
```
